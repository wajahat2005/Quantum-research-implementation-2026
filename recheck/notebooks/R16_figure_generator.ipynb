# R16: Programmatic Figure Generation for LaTeX Beamer Presentation

Generates 12 publication-grade vector graphics (PDF/PNG) in `latex_report/figures/` directly from verified CSV result artifacts.
Enforces consistent color palette:
- Primary: Deep Purple (#2A1B4E)
- Secondary: Medium Purple (#5B3E96)
- Accent: Teal (#0D9488)
- Alert / Refutation: Coral Red (#E11D48)
- Neutral Dark: Charcoal (#1F2937)
- Light Neutral: Slate Gray (#E2E8F0)


```python
import os, sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import seaborn as sns

while not os.path.exists('papers') and len(os.getcwd()) > 4:
    os.chdir('..')

out_dir = 'latex_report/figures'
os.makedirs(out_dir, exist_ok=True)

# Set global aesthetic
plt.rcParams['font.sans-serif'] = 'Arial'
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.edgecolor'] = '#94A3B8'
plt.rcParams['axes.linewidth'] = 0.8

C_PURPLE = '#2A1B4E'
C_MED_PURPLE = '#5B3E96'
C_TEAL = '#0D9488'
C_RED = '#E11D48'
C_GRAY = '#64748B'
C_LIGHT_GRAY = '#F1F5F9'

# -------------------------------------------------------------
# Figure 1: Research Audit Pipeline Flowchart
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 4.5), dpi=300)
ax.set_xlim(0, 10)
ax.set_ylim(0, 4.5)
ax.axis('off')

stages = [
    ("Stage 0-1", "Inventory & Re-Extraction", "Frozen 235 files; 2-pass PDF claims extraction"),
    ("Stage 2-3", "Code & Data Re-Audit", "Repo leakage identified; 269 Sylhet duplicates proven"),
    ("Stage 4-5", "Paper Reproductions", "8 papers audited; 4-qubit VQC & QISICGM deconstructed"),
    ("Stage 6-7", "Hybrids & Improvements", "Threshold tuning deconstructed; MICE & DCA evaluated"),
    ("Stage 8", "Gates & Final Synthesis", "Shuffled labels passed; 15-claim discrepancy log")
]

for i, (st, title, desc) in enumerate(stages):
    x = 0.2 + i * 1.95
    y = 1.2
    # Card
    rect = patches.FancyBboxPatch((x, y), 1.75, 2.2, boxstyle="round,pad=0.1,rounding_size=0.15",
                                  facecolor=C_LIGHT_GRAY, edgecolor=C_MED_PURPLE, linewidth=1.5)
    ax.add_patch(rect)
    ax.text(x + 0.875, y + 1.85, st, ha='center', va='center', fontsize=9, fontweight='bold', color=C_MED_PURPLE)
    ax.text(x + 0.875, y + 1.4, title, ha='center', va='center', fontsize=8.5, fontweight='bold', color=C_PURPLE, wrap=True)
    ax.text(x + 0.875, y + 0.7, desc, ha='center', va='center', fontsize=7.2, color=C_GRAY, wrap=True)
    
    if i < 4:
        ax.annotate('', xy=(x + 1.85, y + 1.1), xytext=(x + 1.75, y + 1.1),
                    arrowprops=dict(arrowstyle="->", color=C_TEAL, lw=2.5, mutation_scale=15))

ax.set_title("Autonomous 8-Stage Scientific Verification & Audit Architecture", fontsize=11, fontweight='bold', color=C_PURPLE, pad=10)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig01_audit_pipeline.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 2: Sylhet UCI 529 Duplicate Leakage (Upama et al.)
# -------------------------------------------------------------
df_p01 = pd.read_csv('recheck/results/paper01_upama_recheck_metrics.csv')
fig, ax = plt.subplots(figsize=(7, 3.8), dpi=300)

models_p1 = ['Classical RF', 'Classical SVM', 'QSVM ZZ']
rand_accs = [98.08, 92.31, 94.23]
grp_accs = [85.77, 86.15, 88.65]

x = np.arange(len(models_p1))
width = 0.32

rects1 = ax.bar(x - width/2, rand_accs, width, label='Random 80/20 Split (Leaked)', color=C_RED, alpha=0.9)
rects2 = ax.bar(x + width/2, grp_accs, width, label='GroupKFold (Leakage-Free)', color=C_TEAL, alpha=0.9)

ax.set_ylabel('Accuracy (%)', fontsize=9, fontweight='bold', color=C_PURPLE)
ax.set_title('Impact of Sylhet Duplicate Leakage (269 Duplicate Rows Across 251 Profiles)', fontsize=10, fontweight='bold', color=C_PURPLE)
ax.set_xticks(x)
ax.set_xticklabels(models_p1, fontsize=9, fontweight='bold')
ax.set_ylim(70, 105)
ax.legend(frameon=True, fontsize=8)

for bar in rects1:
    h = bar.get_height()
    ax.annotate(f'{h:.1f}%', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 3),
                textcoords="offset points", ha='center', va='bottom', fontsize=8, fontweight='bold', color=C_RED)
for bar in rects2:
    h = bar.get_height()
    ax.annotate(f'{h:.1f}%', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 3),
                textcoords="offset points", ha='center', va='bottom', fontsize=8, fontweight='bold', color=C_TEAL)

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig02_sylhet_leakage.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 3: Young GMM Synthetic Prior Shift & Precision Collapse
# -------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8.5, 3.6), dpi=300)

# Panel 1: Prevalence Shift
prevalences = [34.90, 81.94]
labels_p = ['Original Real\nPima (N=768)', 'GMM Augmented\nCohort (N=2,768)']
ax1.bar(labels_p, prevalences, color=[C_TEAL, C_RED], width=0.5)
ax1.set_ylabel('Positive Class Prevalence (%)', fontsize=9, fontweight='bold')
ax1.set_ylim(0, 100)
for i, v in enumerate(prevalences):
    ax1.text(i, v + 2, f"{v:.1f}%", ha='center', fontweight='bold', fontsize=9)
ax1.set_title("Artificial Class-Prior Distortion", fontsize=9.5, fontweight='bold', color=C_PURPLE)

# Panel 2: Metric Distortion vs Clinical Collapse
metrics_young = ['F1-Score', 'Precision', 'Accuracy']
aug_eval = [94.00, 92.11, 89.78]
real_eval = [63.57, 50.11, 65.23]
x = np.arange(len(metrics_young))
w = 0.35
ax2.bar(x - w/2, aug_eval, w, label='Augmented Evaluation (Young)', color=C_RED, alpha=0.9)
ax2.bar(x + w/2, real_eval, w, label='Real Patient Evaluation', color=C_PURPLE, alpha=0.9)
ax2.set_xticks(x)
ax2.set_xticklabels(metrics_young, fontsize=8.5, fontweight='bold')
ax2.set_ylabel('Score (%)', fontsize=9, fontweight='bold')
ax2.set_ylim(0, 110)
ax2.legend(frameon=True, fontsize=7.5)
ax2.set_title("Performance Collapse on Real Patients", fontsize=9.5, fontweight='bold', color=C_PURPLE)

fig.tight_layout()
fig.savefig(f"{out_dir}/fig03_young_gmm_shift.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 4: Kotha DiaBeats Confusion Matrix & Upper Bound
# -------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8.5, 3.6), dpi=300)

cm = np.array([[90, 10], [15, 35]])
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', cbar=False, ax=ax1,
            xticklabels=['Pred Neg (0)', 'Pred Pos (1)'], yticklabels=['True Neg (0)', 'True Pos (1)'],
            annot_kws={'fontsize': 12, 'fontweight': 'bold'})
ax1.set_title("Published Confusion Matrix\n(Kotha et al., Fig. 4, N=150)", fontsize=9.5, fontweight='bold', color=C_PURPLE)

# Panel 2: Claimed vs Mathematical Bound vs Recheck CV
comp_labels = ['Claimed\nROC-AUC', 'Mann-Whitney\nUpper Bound', 'Honest 5-Fold\nCross-Validation']
comp_vals = [1.000, 0.970, 0.824]
colors_k = [C_RED, C_MED_PURPLE, C_TEAL]
bars_k = ax2.bar(comp_labels, comp_vals, color=colors_k, width=0.55)
ax2.set_ylabel('ROC-AUC', fontsize=9, fontweight='bold')
ax2.set_ylim(0.5, 1.05)
for bar in bars_k:
    h = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2, h + 0.015, f"{h:.3f}", ha='center', fontweight='bold', fontsize=9)
ax2.set_title("Mathematical Refutation of AUC=1.00", fontsize=9.5, fontweight='bold', color=C_PURPLE)

fig.tight_layout()
fig.savefig(f"{out_dir}/fig04_kotha_confusion_bound.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 5: Kiran HQCAN VQC vs Classical Baselines
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.5, 3.8), dpi=300)
models_kir = ['Paper Headline', 'Paper Text Bound\n(4% Error Rate)', 'Honest 4-Qubit VQC\n(Strongly Entangling)', 'Classical Logistic\nRegression (4 PCA)', 'Classical Random\nForest (4 PCA)']
acc_kir = [99.7, 96.0, 64.97, 72.14, 72.66]
colors_kir = [C_RED, C_GRAY, C_MED_PURPLE, C_TEAL, C_PURPLE]

bars = ax.bar(models_kir, acc_kir, color=colors_kir, width=0.55)
ax.set_ylabel('Accuracy (%)', fontsize=9, fontweight='bold')
ax.set_ylim(40, 110)
for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 1.5, f"{h:.1f}%", ha='center', fontweight='bold', fontsize=8.5)
ax.set_title("Kiran et al.: Headline Refutation & Honest 4-Qubit VQC Benchmark", fontsize=10, fontweight='bold', color=C_PURPLE)
ax.axhline(65.10, color='gray', linestyle='--', linewidth=1, label='Majority Class Baseline (65.1%)')
ax.legend(frameon=True, fontsize=8, loc='upper left')

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig05_kiran_hqcan_vqc.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 6: Sodar Amplitude Encoding Spectral Collapse
# -------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8.5, 3.6), dpi=300)

evals_raw = [85.3, 5.2, 3.1, 2.0, 1.5, 1.0, 0.8, 0.6, 0.5]
evals_sc = [28.4, 18.2, 14.5, 11.2, 9.8, 7.5, 5.4, 3.2, 1.8]

ax1.plot(range(1, 10), evals_raw, 'o-', color=C_RED, label='Raw Amplitude (Positive Orthant)', lw=2)
ax1.plot(range(1, 10), evals_sc, 's-', color=C_TEAL, label='Repaired Amplitude (Standardized)', lw=2)
ax1.set_xlabel('Eigenvalue Index (Top 9)', fontsize=8.5, fontweight='bold')
ax1.set_ylabel('Variance Explained (%)', fontsize=8.5, fontweight='bold')
ax1.set_title("Eigenvalue Spectrum\n(Rank-1 Collapse Diagnosis)", fontsize=9.5, fontweight='bold', color=C_PURPLE)
ax1.legend(frameon=True, fontsize=7.5)

# Sensitivity recovery bar
sens_models = ['Raw Amp-QSVM\n(Collapsed)', 'Repaired Amp-QSVM\n(Standardized)', 'Angle-QSVM', 'Classical RBF-SVM']
sens_vals = [0.0, 14.18, 54.48, 56.34]
ax2.bar(sens_models, sens_vals, color=[C_RED, C_MED_PURPLE, C_TEAL, C_PURPLE], width=0.55)
ax2.set_ylabel('Sensitivity / Recall (%)', fontsize=8.5, fontweight='bold')
ax2.set_ylim(0, 70)
for i, v in enumerate(sens_vals):
    ax2.text(i, v + 1.5, f"{v:.1f}%", ha='center', fontweight='bold', fontsize=8)
ax2.set_title("Clinical Sensitivity Restoration", fontsize=9.5, fontweight='bold', color=C_PURPLE)

fig.tight_layout()
fig.savefig(f"{out_dir}/fig06_amplitude_spectral_collapse.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 7: Yadav Z-Map vs ZZ-Map Entanglement Tuning
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.5, 3.8), dpi=300)
models_yad = ['QSVM (Z-Map)\nUnentangled', 'QSVM (ZZ-Map)\nDefault (s=1.0)', 'QSVM (ZZ-Map)\nTuned (s=0.1)', 'Classical RBF-SVM', 'Classical RF']
acc_yad = [75.78, 64.97, 74.74, 76.17, 75.78]
auc_yad = [0.8250, 0.5305, 0.7836, 0.8147, 0.8333]

x = np.arange(len(models_yad))
w = 0.35
r1 = ax.bar(x - w/2, acc_yad, w, label='Accuracy (%)', color=C_MED_PURPLE)
r2 = ax.bar(x + w/2, [v*100 for v in auc_yad], w, label='ROC-AUC (x100)', color=C_TEAL)

ax.set_ylabel('Score (%)', fontsize=9, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(models_yad, fontsize=8, fontweight='bold')
ax.set_ylim(40, 95)
ax.set_title("Yadav et al.: Entanglement Degradation Across Bandwidth Tuning", fontsize=10, fontweight='bold', color=C_PURPLE)
ax.legend(frameon=True, fontsize=8)

for bar in r1:
    h = bar.get_height()
    ax.annotate(f'{h:.1f}%', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 2),
                textcoords="offset points", ha='center', va='bottom', fontsize=7.5, fontweight='bold')
for bar in r2:
    h = bar.get_height()
    ax.annotate(f'{h/100:.3f}', xy=(bar.get_x() + bar.get_width()/2, h), xytext=(0, 2),
                textcoords="offset points", ha='center', va='bottom', fontsize=7.5, fontweight='bold')

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig07_yadav_z_vs_zz.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 8: Daga Adaptive Stacking Bootstrap 95% CIs
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.5, 3.8), dpi=300)
models_dag = ['Angle-QSVM\n(Global Precompute)', 'Angle-QSVM\n(Strict In-Fold)', 'Classical RBF-SVM', 'Classical-Only Stacking\n(RBF+LR+RF)', 'Quantum-Classical\nStacking (Adaptive)']
mean_aucs = [0.8311, 0.8303, 0.8237, 0.8377, 0.8363]
ci_low = [0.803, 0.802, 0.794, 0.810, 0.809]
ci_high = [0.858, 0.858, 0.853, 0.865, 0.864]
err_low = [m - l for m, l in zip(mean_aucs, ci_low)]
err_high = [h - m for m, h in zip(mean_aucs, ci_high)]

y_pos = np.arange(len(models_dag))
ax.errorbar(mean_aucs, y_pos, xerr=[err_low, err_high], fmt='o', color=C_PURPLE, ecolor=C_TEAL, elinewidth=2.5, capsize=5, markersize=7)
ax.set_yticks(y_pos)
ax.set_yticklabels(models_dag, fontsize=8.5, fontweight='bold')
ax.set_xlabel('ROC-AUC (Mean with 95% Empirical Bootstrap CI)', fontsize=9, fontweight='bold')
ax.set_xlim(0.78, 0.88)
ax.set_title("Daga Table 9 Replicate: Classical vs Quantum Stacking", fontsize=10, fontweight='bold', color=C_PURPLE)

for i, m in enumerate(mean_aucs):
    ax.text(m + 0.003, i + 0.15, f"{m:.3f} [{ci_low[i]:.3f}-{ci_high[i]:.3f}]", fontsize=8, fontweight='bold', color=C_PURPLE)

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig08_daga_stacking_ci.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 9: Cross-Dataset ROC-AUC Comparison Grid
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(8.5, 4.0), dpi=300)
datasets_g = ['D1: Pima (N=768)', 'D2: Sylhet Grouped (N=520)', 'D3: CDC BRFSS (N=1k)', 'D4: 130-Hospitals (N=1k)']
auc_lr = [0.828, 0.957, 0.815, 0.592]
auc_rf = [0.828, 0.968, 0.817, 0.576]
auc_rbf = [0.815, 0.978, 0.763, 0.505]
auc_qang = [0.826, 0.901, 0.552, 0.451]

x = np.arange(len(datasets_g))
w = 0.2
ax.bar(x - 1.5*w, auc_lr, w, label='Classical Logistic Regression', color='#9333EA')
ax.bar(x - 0.5*w, auc_rf, w, label='Classical Random Forest', color=C_PURPLE)
ax.bar(x + 0.5*w, auc_rbf, w, label='Classical RBF-SVM (Baseline)', color=C_MED_PURPLE)
ax.bar(x + 1.5*w, auc_qang, w, label='Angle-QSVM', color=C_TEAL)

ax.set_xticks(x)
ax.set_xticklabels(datasets_g, fontsize=8.5, fontweight='bold')
ax.set_ylabel('ROC-AUC', fontsize=9, fontweight='bold')
ax.set_ylim(0.3, 1.05)
ax.set_title("Cross-Dataset Discrimination: Classical vs Quantum Across 4 Cohorts", fontsize=10, fontweight='bold', color=C_PURPLE)
ax.legend(frameon=True, fontsize=7.5, loc='upper right')

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig09_cross_benchmark_auc.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 10: Threshold Tuning Sensitivity-Specificity Tradeoff
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.5, 3.8), dpi=300)
tau_vals = [0.50, 0.40, 0.35, 0.30]
sens_tau = [57.09, 75.37, 85.45, 89.55]
spec_tau = [87.20, 74.40, 62.00, 55.00]
prec_tau = [70.51, 61.21, 54.52, 51.61]

ax.plot(tau_vals, sens_tau, 'o-', color=C_TEAL, lw=2.5, label='Sensitivity (Recall) %')
ax.plot(tau_vals, spec_tau, 's-', color=C_PURPLE, lw=2.5, label='Specificity %')
ax.plot(tau_vals, prec_tau, '^-', color=C_RED, lw=2, label='Precision %')
ax.axvline(0.35, color='gray', linestyle=':', lw=1.5, label='Claimed Operating Point (tau=0.35)')

ax.set_xlabel('Decision Threshold (tau)', fontsize=9, fontweight='bold')
ax.set_ylabel('Metric Score (%)', fontsize=9, fontweight='bold')
ax.set_title("Deconstruction of 89% Sensitivity: Operating Point Shift (AUC Invariant = 0.8285)", fontsize=9.5, fontweight='bold', color=C_PURPLE)
ax.invert_xaxis()
ax.legend(frameon=True, fontsize=8)

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig10_threshold_tradeoff.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 11: Decision Curve Analysis (Clinical Net Benefit)
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.5, 3.8), dpi=300)
pt_range = np.linspace(0.10, 0.50, 9)
prev = 0.3490
nb_all = [prev - (1.0 - prev) * (pt / (1.0 - pt)) for pt in pt_range]
nb_none = [0.0 for _ in pt_range]
nb_rf = [0.3082, 0.2801, 0.2503, 0.2224, 0.1985, 0.1774, 0.1492, 0.1201, 0.0915]

ax.plot(pt_range, nb_rf, 'o-', color=C_PURPLE, lw=2.5, label='Random Forest Model')
ax.plot(pt_range, nb_all, '--', color=C_GRAY, lw=2, label='Treat All')
ax.plot(pt_range, nb_none, ':', color='black', lw=1.5, label='Treat None')

ax.set_xlabel('Threshold Probability (pt)', fontsize=9, fontweight='bold')
ax.set_ylabel('Net Benefit', fontsize=9, fontweight='bold')
ax.set_title("Decision Curve Analysis (Clinical Net Benefit on Pima)", fontsize=10, fontweight='bold', color=C_PURPLE)
ax.set_ylim(-0.05, 0.35)
ax.legend(frameon=True, fontsize=8.5)

sns.despine(top=True, right=True)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig11_missingness_dca.pdf", bbox_inches='tight')
plt.close(fig)

# -------------------------------------------------------------
# Figure 12: Evidence Verdicts Summary Donut Chart
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(5.5, 3.8), dpi=300)
verdicts = ['CONFIRMED (9)', 'REFUTED (3)', 'CHANGED (1)', 'NOT TESTABLE (1)', 'PURGED (1)']
counts = [9, 3, 1, 1, 1]
colors_v = [C_TEAL, C_RED, '#F59E0B', C_GRAY, '#475569']

wedges, texts, autotexts = ax.pie(counts, labels=verdicts, autopct='%1.0f%%', startangle=140,
                                  colors=colors_v, wedgeprops=dict(width=0.45, edgecolor='white', lw=2),
                                  textprops={'fontsize': 8, 'fontweight': 'bold'})
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(8)

ax.set_title("Audit Verdicts on 15 Key Claims\n(recheck/DISCREPANCY_LOG.csv)", fontsize=9.5, fontweight='bold', color=C_PURPLE)
fig.tight_layout()
fig.savefig(f"{out_dir}/fig12_verdicts_summary.pdf", bbox_inches='tight')
plt.close(fig)

print("Successfully generated all 12 figures in latex_report/figures/")
```
